# SIMPLE LLM CALLS WITH STREAMING

```python
import langchain
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
# SIMPLE LLM CALL

# INITIALIZING LLM MODEL

from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
# model = init_chat_model("groq/openai/etc:llmname")      # generic code to initialize the model

model = init_chat_model("groq:openai/gpt-oss-120b")
model
```

```python
# Second way to initialize the model

from langchain_groq import ChatGroq
llm = ChatGroq(model="groq:openai/gpt-oss-120b")
llm
```

my LLM has been created till now

```python
# Create Messages

messages = [

    SystemMessage("You are a helpful AI Assisstant"),
    HumanMessage("What are top 2 benefits of using langchain")
]
```

```python
# To pass this message we invoke the model

response = model.invoke(messages)
response

```

```python
print(response.content)
```

```python
# you can directly invoke

model.invoke([HumanMessage("What is machine learning")])
```

```python
# Streaming Example

for chunk in model.stream(messages):
    print(chunk.content, end="", flush=True)
```

# Dynamic Prompt Templates

```python
from langchain_core.prompts import ChatPromptTemplate

# Create Translation App

translation_template = ChatPromptTemplate.from_messages([
    ("system", "You are a professional translator. Translate the follow text {text} from {source_language} to {target_language}. Maintain the tone and style"),
    ("user","{text}")
])

# Using the template

prompt=translation_template.invoke({
    "source_language":"English",
    "target_language":"Sanskrit",
    "text":"Langchain makes building AI applications incredibly easy"
})
```

```python
prompt
```

```python
translated_response = model.invoke(prompt)
print(translated_response.content)
```
